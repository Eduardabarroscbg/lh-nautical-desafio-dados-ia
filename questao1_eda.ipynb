## Parte 3 — Interpretação

A tabela `orders` no total são 48.998 pedidos e 13 colunas, cobrindo de 01/01/2020 até 31/12/2026 — 
bate com o período que a empresa falou que registra, então já é um bom sinal.

Fui atrás de duplicatas e não achei nenhuma (nem no `id`, nem linha repetida 
inteira), o que é ótimo porque significa que ninguém importou o mesmo pedido 
duas vezes por engano. Também conferi se a conta batia (subtotal - desconto = 
total) e deu certo na maioria esmagadora dos casos, então não vi motivo pra 
desconfiar dos valores.

O único ponto que chamou minha atenção foi o `salesperson_id`, vazio em quase 
metade dos pedidos (~49%). No começo fiquei meio desconfiado, mas quando fui 
olhar por canal de venda percebi que faz sentido: pedido de e-commerce não tem 
vendedor físico associado, então é esperado que fique nulo.

Sobre a coluna `total`, os valores vão de R$ 32,62 até R$ 127.262,02, com média 
de R$ 28.704,99 e mediana de R$ 25.917,84. Essa diferença entre média e mediana 
já mostra que a distribuição não é simétrica — tem alguns pedidos bem mais caros 
puxando a média pra cima, o que é normal em dados de venda.

**Minha conclusão:** os dados de `orders` parecem confiáveis — não achei 
duplicata, nem inconsistência nos valores, e os nulos que existem fazem sentido 
de negócio. O único cuidado que eu teria antes de calcular faturamento é olhar 
os pedidos com status `cancelled` e `draft`, que provavelmente não deveriam 
entrar em métricas de receita.
